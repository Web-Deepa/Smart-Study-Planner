## 1.2 Numbers and Strings

### Theory
In Python, data comes in different types. The most basic types are numbers and strings (text).

### Numbers

**Syntax:**
```python
# Integers (whole numbers)
variable_name = 123

# Floats (decimal numbers)  
variable_name = 12.34
```

````python
# Integer examples
age = 25
year = 20.24
negative_number = -10

print("Age:", age)
print("Year:", year)
print("Negative:", negative_number)
````

````python
# Float examples
height = 5.9
price = 19.99
temperature = -3.5

print("Height:", height)
print("Price:", price)
print("Temperature:", temperature)
````

````python
# You can check the type of a variable
print("Type of age:", type(age))
print("Type of height:", type(height))
````

### Strings

**Theory:** Strings are sequences of characters (letters, numbers, symbols) used to represent text.

**Syntax:**
```python

# Double quotes

# Triple quotes (for multi-line)
variable_name = """multiple
lines of text"""
```

````python
# String examples
first_name = 'Ram'
last_name = "Sharma"
city = "Dhangadi"

print("First name:", first_name)
print("Last name:", last_name)
print("City:", city)
````

````python
# Multi-line strings
address = """Dhangadi
Kailali, Sudurpaschim Province  
Nepal"""

print("Address:")
print(address)
````

````python
# String with numbers (still a string!)
phone = "987-6543210"
print("Phone:", phone)
print("Type of phone:", type(phone))
````

## 1.3 Comments

### Theory
Comments are notes in your code that Python ignores when running the program. They help you and others understand what your code does.

### Single-line Comments

**Syntax:**
```python

```

````python
# Calculate the area of a rectangle
length = 10  # length in meters
width = 5   # width in meters
area = length * width  # multiply length and width

print("Area:", area)  # display the result
````

## 1.4 Variables

### Theory
Variables are containers that store data values. Think of them as labeled boxes where you can put information and retrieve it later.

**Rules for Variable Names:**
- Must start with a letter or underscore (_)
- Can contain letters, numbers, and underscores
- Case-sensitive (age and Age are different)
- Cannot use Python keywords (like print, if, for)

### Variable Assignment

**Syntax:**
```python
variable_name = value
```

````python
# Basic assignment
student_name = "Emma"
student_age = 19
is_student = True

print("Name:", student_name)
print("Age:", student_age)
print("Is student:", is_student)
````

````python
# Variables can be reassigned
student_age = 20  # Emma had a birthday!
print("New age:", student_age)
````

````python
# Same value to multiple variables
a = b = c = 10
print("a =", a, "b =", b, "c =", c)
````

````python
# Same value to multiple variables
a = b = c = 10
print("a =", a, "b =", b, "c =", c)
````

### Variable Naming Conventions
```python
# Good variable names (descriptive and clear)
student_count = 25
max_temperature = 98.6
user_email = "user@example.com"

# Poor variable names (unclear)
s = 25  # What does 's' represent?
temp = 98.6  # Temperature? Temporary?
e = "user@example.com"  # Not descriptive

# Use snake_case for variable names
first_name = "John"  # Good
firstName = "John"   # Works but not Python convention
```

## 1.6 Arithmetic Operators

### Theory
Arithmetic operators perform mathematical operations on numbers.

### Basic Operators

**Syntax & Examples:**

````python
# Addition (+)
result = 5 + 3  # 8
print("5 + 3 =", result)
````

````python
# Subtraction (-)
result = 10 - 4  # 6
print("10 - 4 =", result)
````

````python
# Multiplication (*)
result = 6 * 7  # 42
print("6 * 7 =", result)
````

````python
# Division (/)
result = 15 / 4  # 3.75 (always returns float)
print("15 / 4 =", result)
````

````python
# Floor Division (//) - rounds down to nearest integer
result = 15 // 4  # 3
print("15 // 4 =", result)
````

````python
# Modulus (%) - remainder after division
result = 15 % 4  # 3
print("15 % 4 =", result)
````

````python
# Exponentiation (**)
result = 2 ** 3  # 8 (2 to the power of 3)
print("2 ** 3 =", result)
````

### Working with Variables

````python
# Using variables in calculations
a = 10
b = 3

print("a =", a, "b =", b)
print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a // b =", a // b)
print("a % b =", a % b)
print("a ** b =", a ** b)
````

````python
# Updating variables
a = a + 5  # a becomes 15
print("After a = a + 5, a =", a)
````

````python
# Shorthand operators
a += 5  # Same as a = a + 5
print("After a += 5, a =", a)
````

## 1.7 Data Conversion (Type Casting)

### Theory
Sometimes you need to convert data from one type to another. This process is called type casting or type conversion.

### Conversion Functions

**Syntax:**
```python
int(value)    # Convert to integer
float(value)  # Convert to float
str(value)    # Convert to string
bool(value)   # Convert to boolean
```
**Examples:**

````python
# String to number conversion
age_string = "25"
age_number = int(age_string)
print("String:", age_string, "Type:", type(age_string))
print("Integer:", age_number, "Type:", type(age_number))
````

````python
# Number to string conversion
score = 95
score_string = str(score)
print("Number:", score, "Type:", type(score))
print("String:", score_string, "Type:", type(score_string))
````

````python
# Float conversions
price_string = "19.99"
price_float = float(price_string)
print("Price as float:", price_float)
````

````python
# Integer to float
whole_number = 10
decimal_number = float(whole_number)
print("Integer:", whole_number, "Float:", decimal_number)
````

````python
# Float to integer (truncates decimal part)
decimal = 9.8
whole = int(decimal)
print("Float:", decimal, "Integer:", whole)
````

### Boolean Conversion

````python
# Numbers to boolean
print("bool(0):", bool(0))      # False
print("bool(1):", bool(1))      # True
print("bool(-5):", bool(-5))    # True
print("bool(0.0):", bool(0.0))  # False

# Strings to boolean
print("bool(''):", bool(''))       # False (empty string)
print("bool('Hello'):", bool('Hello'))  # True
print("bool('0'):", bool('0'))     # True (non-empty string)

# Boolean to number
print("int(True):", int(True))   # 1
print("int(False):", int(False)) # 0
````

### Practical Examples

````python
# Calculator example
print("=== Simple Calculator ===")
num1_str = "10"  # Imagine this came from user input
num2_str = "5"   # Imagine this came from user input

# Convert strings to numbers
num1 = int(num1_str)
num2 = int(num2_str)

# Perform calculations
print(f"Addition: {num1} + {num2} = {num1 + num2}")
print(f"Multiplication: {num1} × {num2} = {num1 * num2}")

# Age calculator
birth_year_str = "2000"
current_year = 2024
birth_year = int(birth_year_str)
age = current_year - birth_year
print(f"If you were born in {birth_year}, you are {age} years old")
````

## 1.8 Get User Input

### Theory
The input() function allows your program to interact with users by accepting text input. All input is received as a string, regardless of what the user types.

### Basic Input

**Syntax:**
```python
variable = input("Prompt message: ")
```

**Examples:**

````python
# Basic input
name = input("What is your name? ")
print("Hello,", name)

# Input with processing
age_str = input("How old are you? ")
age = int(age_str)  # Convert string to integer
print(f"Next year you will be {age + 1} years old")

# Multiple inputs
first_name = input("First name: ")
last_name = input("Last name: ")
print(f"Full name: {first_name} {last_name}")
````

### Input with Different Data Types

````python
# Getting numbers from user
print("=== Circle Area Calculator ===")
radius_str = input("Enter the radius: ")
radius = float(radius_str)  # Convert to float for decimal numbers
pi = 3.14159
area = pi * radius ** 2
print(f"The area of a circle with radius {radius} is {area:.2f}")

# Getting multiple pieces of information
print("\n=== Student Registration ===")
student_name = input("Student name: ")
student_age = int(input("Student age: "))
student_gpa = float(input("Student GPA: "))

print(f"\nStudent Information:")
print(f"Name: {student_name}")
print(f"Age: {student_age}")
print(f"GPA: {student_gpa}")
````

### Common Errors & Debugging:
```python
# ERROR: Forgetting to convert input
# age = input("Your age: ")
# next_year = age + 1  # TypeError: can't add string and int

# CORRECT:
age = int(input("Your age: "))
next_year = age + 1

# ERROR: Invalid conversion
# This will crash if user enters non-numeric input
# number = int(input("Enter a number: "))  # ValueError if user enters "abc"

# BETTER: Add error handling

    number = int(input("Enter a number: "))
    print("You entered:", number)
```
